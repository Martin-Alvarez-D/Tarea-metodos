```python
#Tarea 2.10, Martín Alvarez

import numpy as np
import matplotlib.pyplot as plt
true = np.sqrt(2)
x = 2.0
k = 0
N = 20

x_new_hist=[]
error_hist=[]

while k<N:
    x_new=0.5*(x+2/x)
    x_new_hist.append(x_new)
    x=x_new
    k=k+1
    error=abs(x_new-true)
    error_hist.append(error)

f = lambda x: x**2 - 2.0

# Método de bisección
a, b = 1.0, 2.0
err_bis = np.empty(N)
x_bis_hist = []

for k in range(N):
    c = 0.5*(a + b)
    x_bis_hist.append(c)
    err_bis[k] = abs(c - true)
    if f(a)*f(c) <= 0:
        b = c
    else:
        a = c

plt.semilogy(range(N), error_hist, label="Newton")
plt.semilogy(range(N), err_bis, label="Bisección")
plt.xlabel("Iteración")
plt.ylabel("Error absoluto")
plt.legend()
plt.show()

#El metodo de Newton disminuye mas rapido que el de bisección, sin embargo su estancamiento ocurre ya que el computador cuenta con una capacidad finita de precisión con la cual puede representar los números
```
