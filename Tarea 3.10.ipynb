```python
#Tare 3.10 Martin Alvarez

import numpy as np
import matplotlib.pyplot as plt

#Particulas en 10 segundos
16.67

#para extraer el numero de particulas con poisson
numero_de_particulas=np.random.poisson(16.67)

media=0.99
error=0.01
cantidad=numero_de_particulas

#con esto calculamos las velocidades para las particulas detectadas
velocidades=np.random.normal(media,error,cantidad)

plt.hist(velocidades)
plt.xlabel("velocidad simulada(c)")
plt.ylabel("numero de particulas")
plt.show()
```
