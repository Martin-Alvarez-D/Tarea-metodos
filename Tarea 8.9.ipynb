```python
#Tarea 8.9 Martín Alvarez

# la ecuación a usar: (dv/dt)=-g-(k/m)*v*|v| con v(0)=8

#|v| sirve para la subida y la bajada del proyectil

import numpy as np

def f(t, v):
    g = 9.8
    k = 0.002
    m = 0.11
    return -g-(k/m)*v*np.abs(v)


def euler(f, a, b, yo, N):

    h = (b-a)/N
    t = a
    w = yo
    ts = np.zeros(N+1)
    sol = np.zeros(N+1)

    for i in range(N):
        ts[i] = t
        sol[i] = w
        w = w + h*f(t, w)
        t = t + h

    ts[N] = t
    sol[N] = w

    return ts, sol

ts, vs = euler(f, 0, 1, 8, 100)

print(ts[10::10])
print(vs[10::10])

# t de altura maxima app 0.8
```
