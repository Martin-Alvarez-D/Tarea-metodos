```python
#Tarea 4.1 Martin Alvarez

import numpy as np
import matplotlib.pyplot as plt

#media E[y]=E[Ax+b] ; covarianza E[(y-E[y])(y-E[y])^T]
#E[y]=(6,12) ; Cov(Y)=E[A(x-mu)(A(x-mu))^T]
#            ; Cov(y)=AE[(x-mu)(x-mu)^T]AT
#            ; Cov(y)=ACov(x)A^T
#            ; Cov(y)=(14,32,32,74)

t= np.linspace(0,2*np.pi,200)
z= np.vstack([np.cos(t),np.sin(t)])
sigma= np.array([[2,1],[1,2]])
mu= np.array([[1],[2]])

vals, V=np.linalg.eigh(sigma)
x= mu+V@np.diag(np.sqrt(vals))@z

A=np.array([[1,2],[3,4]])
b=np.array([[1],[1]])
y= A@x+b
mu_y=A@mu+b

plt.plot(x[0],x[1],label="contorno de x")
plt.plot(y[0],y[1],label="contorno de y")
plt.plot(mu[0],mu[1],"o")
plt.plot(mu_y[0],mu_y[1],"o")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()

#la transformación traslada la elipse (siendo el causante de esto b), estira y rota a la elipse (siendo A la causante),
#por ultimo el area es multiplicada por un factor 2 extraido del determinante de A y este ultimo debido a ser un determinante
#con simbolo negativo hace evidencia de una reflexion por parte de la treansformación hacia la elipse.
#Centro de la elipse pasó de (1,2) a (6,11)
#los ejes pasaron de +45°;-45° a 66.6°;-223.4°
#cociente de semiejes pasó de aprox. 1.7° a aprox. 25.4°
```
