```python
#Tarea 6.8 Martin Alvarez
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return np.exp((-x**2)/2)*1/np.sqrt(2*np.pi)

def lagrange(x, puntos):
    Lis = np.ones(len(puntos))
    fs = np.ones(len(puntos))
    for i in range(len(puntos)):
        fs[i] = puntos[i][1]
        for j in range(len(puntos)):
            if j != i:
                Lis[i] *= (x - puntos[j][0])/(puntos[i][0] - puntos[j][0])

    return sum(fs*Lis)

puntos=[(xi,f(xi)) for xi in np.linspace(0, 2, 6)]

xs= np.linspace(0, 2, 200)
plt.plot(xs, [lagrange(xi, puntos) for xi in xs], label= "aproximación", linestyle= "--", color= "red")
plt.plot(xs, f(xs), label="real", linewidth= 1)
plt.legend()

P= np.array([lagrange(xi, puntos) for xi in xs])

error= np.abs(f(xs)-P)
print(np.max(error))
print(lagrange(3, puntos), f(3))
print(np.abs(lagrange(3, puntos) - f(3)))

#El polinomio no sirve para x=3 ya que con P(3) y f(3) se genera un error 540 veces respecto al del intervalo [0, 2] y uno de 40 veces respecto al real.
#Considerando que la gaussiana tiende a 0 y no es negativa, y que fuera de los límites del polinomio tiende al ininito negativo/positivo,
#el polinomio no es utilizable para x=3
```
