```python
#Tarea 5.1 Martin Alvarez

import numpy as np

for L in [1.0, 0.5, 0.1, 0.01, 0.001]:
    x= np.linspace(-L/2, L/2, 1000)
    rho= np.ones_like(x)/L
    S= np.trapezoid(-rho*np.log(rho), x)
    print(L, S, np.log(L))

#Cuando L tiende a 0 S tiende al menos infinito porque se está dividiendo el tramo infinitas veces y no es 0 ya que esto solo representa el punto
#donde el tramo coincide con la referencia, sin embargo, esto no implica una "certeza absoluta".
```
