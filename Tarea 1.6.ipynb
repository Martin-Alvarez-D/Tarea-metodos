```python
import numpy as np
import h5py

def conversor_de_archivo(simulacion_txt,simulacion_hd5):
    datos=np.loadtxt(simulacion_txt)
    col1=datos[:,0]
    col2=datos[:,1]
    col3=datos[:,2]
    with h5py.File(simulacion_hd5, "w") as archivo:
        archivo.create_dataset("col1", data=col1)
        archivo.create_dataset("col2", data=col2)
        archivo.create_dataset("col3", data=col3)

conversor_de_archivo("simulacion.txt", "simulacion.hd5")

with h5py.File("simulacion.hd5", "r") as archivo:
    print("Datasets:", list(archivo.keys()))

    print("col1:", archivo["col1"][:])
    print("col2:", archivo["col2"][:])
    print("col3:", archivo["col3"][:])
```
