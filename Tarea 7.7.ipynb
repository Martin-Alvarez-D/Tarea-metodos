```python
#Tarea 7.7 Martin Alvarez

import numpy as np

def gauss_quad_5(f, a, b):

    def x(t):
        return ((b - a)*t + a + b)/2

    c = [0.2369268850, 0.4786286705, 0.5688888889, 0.4786286705, 0.2369268850]
    r = [0.9061798459, 0.5384693101, 0.0000000000, -0.5384693101, -0.9061798459]

    return ((b-a)/2)*sum(c[i]*f(x(r[i])) for i in range(5))
def gauss_compuesta(f, a, b, n):
    h = (b - a) / n
    total = 0
    for i in range(n):
        izquierda = a+i*h
        derecha = a+(i+1)*h
        total += gauss_quad_5(f, izquierda, derecha)
    return total

gauss_compuesta(np.cos, 0, 2, 4)
```
